```python
import os
import cv2
import mediapipe as mp
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import time

VIDEO_FOLDER = './all_videos'
MODEL_PATH = './face_landmarker.task'

FRAME_SKIP = 3
RESIZE_WIDTH = 480

class BlinkAndFaceAnalyzer:
    def __init__(self, video_folder, model_path):
        self.video_folder = video_folder

        self.stats = {
            'total_videos': 0,
            'total_duration': 0,
            'total_blinks': 0,
            'blink_rate': 0,
            'video_details': [],
            'facial_dimensions': None,
            'face_frame': None
        }

        base_options = mp.tasks.BaseOptions(model_asset_path=model_path)
        self.options = mp.tasks.vision.FaceLandmarkerOptions(
            base_options=base_options,
            running_mode=mp.tasks.vision.RunningMode.VIDEO,
            output_face_blendshapes=True,
            num_faces=1
        )

    def get_video_files(self):
        return sorted(Path(self.video_folder).glob('*.MP4'))

    def dist(self, p1, p2):
        return np.linalg.norm(np.array(p1) - np.array(p2))

    def estimate_facial_dimensions(self, lm):

        # ---------- FACE ----------
        face_top = lm[10]
        face_bottom = lm[152]
        face_left = lm[454]
        face_right = lm[234]

        # ---------- NOSE ----------
        nose_top = lm[8]
        nose_bottom = lm[4]
        nose_left = lm[98]
        nose_right = lm[327]

        # ---------- MOUTH ----------
        mouth_left = lm[61]
        mouth_right = lm[291]
        mouth_top = lm[0]
        mouth_bottom = lm[17]

        # ---------- LEFT EYE ----------
        le_left = lm[33]
        le_right = lm[133]
        le_top = lm[159]
        le_bottom = lm[145]

        # ---------- RIGHT EYE ----------
        re_left = lm[362]
        re_right = lm[263]
        re_top = lm[386]
        re_bottom = lm[374]

        # scale (approx 20 cm head height)
        head_h = self.dist(face_top, face_bottom)
        head_w = self.dist(face_left, face_right)
        scale = 20 / head_h if head_h > 0 else 1

        return {
            # HEAD
            'head_height': head_h * scale,
            'head_width': head_w * scale,

            # NOSE
            'nose_height': self.dist(nose_top, nose_bottom) * scale,
            'nose_width': self.dist(nose_left, nose_right) * scale,

            # MOUTH
            'mouth_height': self.dist(mouth_top, mouth_bottom) * scale,
            'mouth_width': self.dist(mouth_left, mouth_right) * scale,

            # LEFT EYE
            'left_eye_width': self.dist(le_left, le_right) * scale,
            'left_eye_height': self.dist(le_top, le_bottom) * scale,

            # RIGHT EYE
            'right_eye_width': self.dist(re_left, re_right) * scale,
            'right_eye_height': self.dist(re_top, re_bottom) * scale,

            # points for drawing
            'points': {
                'face': (face_top, face_bottom, face_left, face_right),
                'nose': (nose_top, nose_bottom, nose_left, nose_right),
                'mouth': (mouth_left, mouth_right, mouth_top, mouth_bottom),
                'left_eye': (le_left, le_right, le_top, le_bottom),
                'right_eye': (re_left, re_right, re_top, re_bottom),
            }
        }

    def process_video(self, video_path, i, total):

        print(f"Processing video {i} of {total}: {video_path.name}")

        cap = cv2.VideoCapture(str(video_path))
        if not cap.isOpened():
            print("Failed")
            return None

        fps = cap.get(cv2.CAP_PROP_FPS)
        frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        duration = frame_count / fps if fps > 0 else 0

        blink_count = 0
        eye_closed = False
        frame_number = 0

        with mp.tasks.vision.FaceLandmarker.create_from_options(self.options) as landmarker:
            while True:
                ret, frame = cap.read()
                if not ret:
                    break

                frame_number += 1
                if frame_number % FRAME_SKIP != 0:
                    continue

                h, w = frame.shape[:2]
                frame = cv2.resize(frame, (RESIZE_WIDTH, int(h * RESIZE_WIDTH / w)))

                rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb)

                ts = int((frame_number / fps) * 1000)
                res = landmarker.detect_for_video(mp_img, ts)

                # BLINK
                if res.face_blendshapes:
                    s = res.face_blendshapes[0]
                    try:
                        l = next(x.score for x in s if x.category_name == 'eyeBlinkLeft')
                        r = next(x.score for x in s if x.category_name == 'eyeBlinkRight')
                        c = (l + r) / 2

                        if c > 0.45:
                            if not eye_closed:
                                blink_count += 1
                                eye_closed = True
                        else:
                            eye_closed = False
                    except:
                        pass

                # capture face once
                if self.stats['facial_dimensions'] is None and res.face_landmarks:
                    lm = [(int(p.x * frame.shape[1]), int(p.y * frame.shape[0]))
                          for p in res.face_landmarks[0]]

                    self.stats['facial_dimensions'] = self.estimate_facial_dimensions(lm)
                    self.stats['face_frame'] = frame.copy()

        cap.release()
        print("Done!")

        return {
            'filename': video_path.name,
            'duration': duration,
            'blinks': blink_count
        }

    def show_face_image(self):

        if self.stats['facial_dimensions'] is None:
            print("No face detected")
            return

        frame = self.stats['face_frame']
        d = self.stats['facial_dimensions']
        p = d['points']

        # ---------- DRAW LINES ----------
        cv2.line(frame, p['face'][0], p['face'][1], (255, 0, 0), 2)
        cv2.line(frame, p['face'][2], p['face'][3], (255, 0, 0), 2)

        cv2.line(frame, p['nose'][0], p['nose'][1], (0, 0, 255), 2)
        cv2.line(frame, p['mouth'][0], p['mouth'][1], (0, 255, 255), 2)

        cv2.line(frame, p['left_eye'][0], p['left_eye'][1], (0, 255, 0), 2)
        cv2.line(frame, p['left_eye'][2], p['left_eye'][3], (0, 255, 0), 2)

        cv2.line(frame, p['right_eye'][0], p['right_eye'][1], (0, 255, 0), 2)
        cv2.line(frame, p['right_eye'][2], p['right_eye'][3], (0, 255, 0), 2)

        # ---------- TEXT ----------
        y = 22
        step = 22

        for text in [
            f"Head H: {d['head_height']:.1f}  W: {d['head_width']:.1f}",
            f"Nose H: {d['nose_height']:.1f}  W: {d['nose_width']:.1f}",
            f"Mouth H: {d['mouth_height']:.1f}  W: {d['mouth_width']:.1f}",
            f"Left Eye: {d['left_eye_width']:.1f} x {d['left_eye_height']:.1f}",
            f"Right Eye: {d['right_eye_width']:.1f} x {d['right_eye_height']:.1f}"
        ]:
            # ---- draw background rectangle ----
            (text_w, text_h), baseline = cv2.getTextSize(
                text, cv2.FONT_HERSHEY_SIMPLEX, 0.45, 1
            )
            cv2.rectangle(
        frame,
        (8, y - text_h - 5),
        (8 + text_w + 6, y + baseline),
        (0, 0, 0),   # black background
        -1           
    )
            cv2.putText(frame, text, (10, y),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.45, (255, 255, 255), 1)
            y += step

        plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.title("Facial Dimensions (in approx cm)")
        plt.show()

    def run_analysis(self):

        videos = self.get_video_files()
        total = len(videos)

        print(f"\nTotal videos in folder: {total}\n")

        start = time.time()

        for i, v in enumerate(videos, 1):
            r = self.process_video(v, i, total)
            if r:
                self.stats['video_details'].append(r)
                self.stats['total_duration'] += r['duration']
                self.stats['total_blinks'] += r['blinks']

        self.stats['blink_rate'] = (
            self.stats['total_blinks'] / self.stats['total_duration']
            if self.stats['total_duration'] > 0 else 0
        )

        runtime = (time.time() - start) / 60
        print(f"\nTotal duration of videos: {self.stats['total_duration']:.2f} sec")
        print(f"Total runtime for analysis: {runtime:.2f} min")

        print("\n========== FINAL RESULTS ==========")
        print(f"Videos: {total}")
        print(f"Blinks: {self.stats['total_blinks']}")
        print(f"Blink Rate: {self.stats['blink_rate']:.2f}/sec")

        self.show_face_image()


def main():
    BlinkAndFaceAnalyzer(VIDEO_FOLDER, MODEL_PATH).run_analysis()


if __name__ == "__main__":
    main()
```
